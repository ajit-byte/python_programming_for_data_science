```python
inventory = {"Widget": 10, "Gadget": 5, "Sensor": 0, "Cable": 15}
```

```python
orders = [
    ["Widget", 3],
    ["Sensor", 2],
    ["Gadget", 7],
    ["Cable", 20],
    ["Drone", 1]
]
```

```python
fulfilled_count = 0
unfulfilled_items = []
```

```python
for order in orders:
    item = order[0]
    requested_qty = order[1]
    stock = inventory.get(item)

    if stock is None:
        print(f"ALERT: '{item}' is not a recognized inventory item.")

    elif stock == 0:
        print(f"OUT OF STOCK: '{item}' currently has 0 units available.")

    elif stock >= requested_qty:
        inventory[item] = stock - requested_qty
        fulfilled_count += 1
        print(f"FULFILLED: Shipped {requested_qty} x {item}. "
              f"Remaining stock: {inventory[item]}.")

    else:
        remainder = requested_qty - stock
        inventory[item] = 0
        unfulfilled_items.append([item, remainder])
        print(f"PARTIAL: Shipped {stock} x {item}, "
              f"{remainder} unit(s) could not be fulfilled.")
```

```python
print("\n--- End of Batch Summary ---")
print("Final Inventory:", inventory)
print("Total Fully Fulfilled Orders:", fulfilled_count)
print("Unfulfilled Items (item, quantity short):", unfulfilled_items)
```

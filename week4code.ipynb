```python

# Task 1: functional tools
def clean_transactions(amounts, min_threshold=10.0):
    # keep only positive amounts that also meet the minimum threshold
    return list(filter(lambda x: x > 0 and x >= min_threshold, amounts))
```

```python

def apply_service_fee(amounts, fee_rate=0.02):
    # deduct the fee percentage from every amount
    return list(map(lambda x: round(x * (1 - fee_rate), 2), amounts))

```

```python

# Task 2: pure recursion
def calculate_nested_total(data):
    # base case 1: empty list
    if data == []:
        return 0.0

    # base case 2: single number
    if isinstance(data, (int, float)):
        return data

    # recursive step: split into head and tail
    head = data[0]
    tail = data[1:]
    return calculate_nested_total(head) + calculate_nested_total(tail)

```

```python

# Testing
raw = [120.50, -30, 5, 0, 45.75, 9.99, 200, 10]
print("Raw transactions:", raw)
```

```python

cleaned = clean_transactions(raw)
print("Cleaned (min 10.0):", cleaned)

```

```python
processed = apply_service_fee(cleaned)
print("After 2% fee:", processed)

print("Cleaned (min 50.0):", clean_transactions(raw, 50.0))
print("After 5% fee:", apply_service_fee(cleaned, 0.05))
```

```python

batch = [100, [50, 25], [[10, 5], 40]]
print("\nNested batch:", batch)
print("Nested total:", calculate_nested_total(batch))
print("Empty list total:", calculate_nested_total([]))
print("Single value total:", calculate_nested_total(75.5))
print("Deep nesting total:", calculate_nested_total([1, [2, [3, [4, [5.5]]]]]))
```

## Create the encrypted input file

```python
raw = open("raw_logs.txt", "w")
raw.write("V|vwhp errw frpsohwhg\n")
raw.write("Xvhu dgplq orjjhg lq\n")
raw.write("EUHDFK ghwhfwhg rq sruw 55\n")
raw.write("Glvn xvdjh dw :; shufhqw\n")
raw.write("Pxowlsoh idlohg orjlqv dqg srvvleoh euhdfk\n")
raw.write("Edfnxs ilqlvkhg vxffhvvixoo|\n")
raw.close()
```

## Task 1: file ingestion and output preparation

```python
infile = open("raw_logs.txt", "r")
master = open("decrypted_master.txt", "w")
alerts = open("security_alerts.txt", "w")
```

## Task 2 and 3: decrypt, Strip, write, and filter

```python
for line in infile:
    decrypted = ""
    for ch in line:
        if ch == " " or ch == "\n":
            decrypted += ch
        else:
            decrypted += chr(ord(ch) - 3)

    decrypted = decrypted.strip()
    master.write(decrypted + "\n")

    if "BREACH" in decrypted.upper():
        alerts.write(decrypted + "\n")
```

## now closing the file

```python
infile.close()
master.close()
alerts.close()
```
